我搭建完整数据分析链路，在代码调试、图表显示、文本排版遇到困难时借助AI优化完善，最终产出符合课程作业标准的完整分析内容。

## 三、各环节实操与AI辅助详情
### 3.1 数据读取与初步理解
导入数据集，自主查看数据行列数量、字段名称，区分户型、小区等文本字段与面积、总价数值，完成基础描述统计查看。
在输出内容杂乱时借助AI精简代码，统一打印格式，让数据概览结果更加清晰直观。

### 3.2 数据质量检查
梳理数据质检要点：缺失值、重复房源、极端异常数据，手动编写部分检测代码，筛查发现部分房源总价、均价字段存在缺失，数据集无重复记录。
AI优化输出逻辑，将缺失字段、缺失数量整理成清晰文字，省去手动整理数据的重复工作。

### 3.3 数据清洗与字段转换
所有清洗分类规则由本人和ai结合二手房业务自行设定：
- 楼层划分为低层、中层、高层；
- 户型按照居室数量分类；
- 总价划分五档预算区间。

我完成字符串提取、分类转换代码，在代码冗余繁琐时使用AI简化写法，清洗规则保留我原定方案。

### 3.4 异常值检验
我选择适配房价偏态分布的IQR四分位法，搭配3σ准则开展双重异常检验，选定面积、总价、均价三个核心数值字段，搭建异常判断基础函数。
绘图出现报错、图片无法展示时，使用AI修复绘图代码，同时优化异常数量、占比汇总表格输出格式。

### 3.5 描述性统计分析
我规划分析维度：不同城区房价对比、楼层价格差异、户型供给分布，手动编写部分分组聚合代码。ai计算各维度均价、中位数等指标，优化数字排版，分析对比结论，结合数据自行总结。

### 3.6 可视化制作
四张图表的分析目的、展示维度借助ai确定并编写基础绘图代码；
运行出现中文方框乱码、画布拥挤、标题模糊等问题时，借助AI补充字体配置、调整画布尺寸、美化图例坐标轴，不改动图表原始数据与核心展示内容。

### 3.7 报告撰写
报告整体框架、各板块核心内容撰写；AI做辅助优化：
1. 统一Markdown标题、列表层级；
2. 把口语化语句调整为规范书面表达；
3. 删减重复段落，梳理行文逻辑；

不新增分析观点，对文本格式、文笔做美化。

## 四、核心分析结论
1. **区域学区是房价第一分层因素**
   支撑依据：行政区分组聚合统计、图表2各行政区均价柱状图。
   核心内容：鼓楼核心学区行政区均价、中位数显著高于仓山、晋安、外围区县，城市二手房区域价格梯度清晰；同时图表1房源数量计数图显示鼓楼挂牌房源总量偏少，学区房供给稀缺。

2. **市场主流需求集中在100-300万中端预算区间**
   支撑依据：总价档位分组统计、图表3预算占比饼图。
   核心内容：100-200万、200-300万房源合计占市场七成以上，是刚需、刚改主力；100万以内小户型、500万以上高端豪宅供给稀缺，可选房源少。

3. **楼层存在稳定价格溢价规律，中层房源价值最高**
   支撑依据：楼层档次分组统计、图表4楼层单价箱线图。
   核心内容：同一行政区内中层房源均价中位数最高，采光、视野综合条件最优；低层房源均价整体最低；高层价格介于中层与低层之间。

4. **3室户型是市场流通主力，小户型学区房单价溢价极强**
   支撑依据：户型室数分组统计、单价TOP10排序结果。
   核心内容：3室房源样本数量最多，供需最旺盛；小户型学区房单位均价远高于大面积改善住宅，学区带来的单价增值超过户型面积带来的总价增值。

5. **刚需自住房源集中在仓山、晋安板块**
   支撑依据：图表1各区域房源数量计数图、行政区分组综合统计。
   核心内容：仓山、晋安挂牌房源数量遥遥领先其他区县，片区户型面积偏大、均价更低，适配纯自住刚需人群。
